```python
# Lista vacía
recepcionista = []

# Diccionario vacío
losrecepcionista = {}

print("===================================")
print(" 🤖 RECEPCIONISTA IA ")
print("===================================")

nombre_usuario = input("Ingresa tu nombre: ")
print("Hola", nombre_usuario, "😊")

while True:
    print("\n----- MENÚ -----")
    print("1. Registrar cliente")
    print("2. Ver lista de clientes")
    print("3. Eliminar último cliente")
    print("4. Salir")

    opcion = int(input("Elige una opción: "))

    if opcion == 1:
        print("\n📋 Nuevo registro")

        losrecepcionista = {}

        losrecepcionista["nombre"] = input("Nombre del cliente: ")
        losrecepcionista["edad"] = input("Edad: ")
        losrecepcionista["motivo"] = input("Motivo de visita: ")

        recepcionista = recepcionista + [losrecepcionista]

        print("✔ Cliente registrado\n")

    elif opcion == 2:
        print("\n📋 LISTA DE CLIENTES:")

        if recepcionista == []:
            print("No hay registros\n")

        else:
            i = 0
            while i < len(recepcionista):
                print("----- Cliente", i + 1, "-----")
                print("Nombre:", recepcionista[i]["nombre"])
                print("Edad:", recepcionista[i]["edad"])
                print("Motivo:", recepcionista[i]["motivo"])
                print("----------------------\n")
                i = i + 1

    elif opcion == 3:
        if recepcionista == []:
            print("❌ No hay clientes para eliminar\n")
        else:
            eliminado = recepcionista[-1]
            recepcionista = recepcionista[:-1]
            print("🗑 Cliente eliminado:", eliminado["nombre"], "\n")

    elif opcion == 4:
        print("👋 Gracias por usar la Recepción IA,", nombre_usuario)
        break

    else:
        print("⚠ Opción no válida\n")
```
